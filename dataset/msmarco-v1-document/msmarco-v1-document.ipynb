## Obtain the corpus

No license agreement required. `ir_datasets` downloads the corpus (~22 GB, 3,213,835 documents) automatically on first access and caches it under `~/.ir_datasets/msmarco-document/`. Download + docstore build take considerably longer than the passage corpus.

```python
import ir_datasets
dataset_name = "msmarco-document"
dataset = ir_datasets.load(dataset_name)
docstore = dataset.docs_store()
docstore.build()
```

Document fields and sample data

```python
print(dataset.docs_cls().__annotations__)
```

Get a sample document (full web pages: `url`, `title`, `body`)

```python
doc_id='D1555982'
doc = docstore.get(doc_id)
print(doc.url)
print(doc.title)
print(doc.body[:500])
```

Corpus size

```python
print(dataset.docs_count())
```

Document length distribution (sample) — drives the encoder-side truncation choice in the indexing notebooks

```python
import itertools
import pandas as pd
lengths = pd.Series([len(d.body) for d in itertools.islice(dataset.docs_iter(), 50_000)])
print(lengths.describe(percentiles=[.5, .75, .9, .99]))
```

## Obtain the queries/qrels

Use the `judged` variants so only queries with NIST judgments are included:

- TREC DL 2019: 43 queries
- TREC DL 2020: 45 queries

```python
dl2019 = ir_datasets.load("msmarco-document/trec-dl-2019/judged")
dl2020 = ir_datasets.load("msmarco-document/trec-dl-2020/judged")
```

Query fields and data

```python
print(dl2019.queries_cls().__annotations__)
```

Show queries

```python
pd.DataFrame(dl2019.queries_iter())
```

```python
pd.DataFrame(dl2020.queries_iter())
```

Qrel fields and data

Qrels are graded (0-3). Binary-metric convention for the document task is relevance >= 1 (vs >= 2 for the passage task).

```python
print(dl2019.qrels_cls().__annotations__)
```

```python
pd.DataFrame(dl2019.qrels_iter())
```

```python
pd.DataFrame(dl2020.qrels_iter())
```

Relevance grade distribution

```python
for name, ds in [('DL 2019', dl2019), ('DL 2020', dl2020)]:
    df = pd.DataFrame(ds.qrels_iter())
    print(name, dict(df['relevance'].value_counts().sort_index()))
```
