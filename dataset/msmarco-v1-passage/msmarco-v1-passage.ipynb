## Obtain the corpus

No license agreement required. `ir_datasets` downloads the corpus (~2.9 GB, 8,841,823 passages) automatically on first access and caches it under `~/.ir_datasets/msmarco-passage/`. Building the docstore takes a while.

```python
import ir_datasets
dataset_name = "msmarco-passage"
dataset = ir_datasets.load(dataset_name)
docstore = dataset.docs_store()
docstore.build()
```

Document fields and sample data

```python
print(dataset.docs_cls().__annotations__)
```

Get a sample document (passages have a single `text` field, no title)

```python
doc_id='7157715'
print(docstore.get(doc_id).text)
```

Corpus size

```python
print(dataset.docs_count())
```

## Obtain the queries/qrels

Use the `judged` variants so only queries with NIST judgments are included:

- TREC DL 2019: 43 queries
- TREC DL 2020: 54 queries

```python
dl2019 = ir_datasets.load("msmarco-passage/trec-dl-2019/judged")
dl2020 = ir_datasets.load("msmarco-passage/trec-dl-2020/judged")
```

Query fields and data

```python
print(dl2019.queries_cls().__annotations__)
```

Show queries

```python
import pandas as pd
pd.DataFrame(dl2019.queries_iter())
```

```python
pd.DataFrame(dl2020.queries_iter())
```

Qrel fields and data

Qrels are graded (0-3): 0 irrelevant, 1 related, 2 highly relevant, 3 perfectly relevant. For binary metrics (e.g. MAP), the track convention is relevance >= 2.

```python
print(dl2019.qrels_cls().__annotations__)
```

```python
pd.DataFrame(dl2019.qrels_iter())
```

```python
pd.DataFrame(dl2020.qrels_iter())
```

Relevance grade distribution

```python
for name, ds in [('DL 2019', dl2019), ('DL 2020', dl2020)]:
    df = pd.DataFrame(ds.qrels_iter())
    print(name, dict(df['relevance'].value_counts().sort_index()))
```
